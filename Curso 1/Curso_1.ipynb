```python
#Funciones
def thing():
    print('Hello')
    print('Fun')

def greet(lang):
    if lang=='es':
        return 'Hola'
    elif lang == 'fr':
        return 'Bonjour'
    else:
        return 'Hello'

thing()
print(greet('es'),'Glenn')
print(greet('uk'),'Sally')
print(greet('fr'),'Michael')
```

```python
#Practica Funciones
def pago_normal(horas,rate):
    pay = horas*rate
    return pay
    
def computepay(horas,rate):
    pay = 40*rate
    pay2 = (horas-40)*rate*1.5
    final_pay = pay + pay2
    return final_pay

horas = float(input('Ingrese las horas:'))
rate = float(input('Ingrese el pago x hora:'))

if horas>=40:
    x = computepay(horas,rate)
    print('Pay',x)
else:
    y = pago_normal(horas,rate)
    print('Pay',y)
```

```python
#Número más grande
mas_grande_que = -1
print('Antes:',mas_grande_que)
for el_mas in [9,41,12,3,74,15]:
    if el_mas > mas_grande_que:
        mas_grande_que = el_mas
    print(mas_grande_que, el_mas)
print('Despues:',mas_grande_que)
```

```python
#Suma en Bucle
zork=0
print('Antes:',zork)
for i in [9,41,12,3,74,15]:
    zork = zork+i
    print(zork,i)
print('Despues:',zork)
```

```python
#Número más pequeño
mas_pequeño_que = None
print('Antes:',mas_pequeño_que)
for el_mas in [9,41,12,3,74,15]:
    if mas_pequeño_que is None:
        mas_pequeño_que = el_mas
    elif el_mas<mas_pequeño_que:
        mas_pequeño_que = el_mas
    print(mas_pequeño_que, el_mas)
print('Despues:',mas_pequeño_que)
```

```python
#Practica Bucles
mayor = None
menor = None

while True:
    try:
        numero = input('Ingrese un numero: ')

        if numero == 'done':
            break

        numero = int(numero)

        if mayor is None or numero > mayor:
            mayor = numero

        if menor is None or numero < menor:
            menor = numero

    except:
        print('Invalid input')

print('Mayor:', mayor)
print('Menor:', menor)
```
